```

import pandas as pd
link_csv = "https://raw.githubusercontent.com/joshirdzdel-svg/estadisticainferencial2026A/refs/heads/main/Unidad2/gym_members/houses_price/Casa_toluca%20(1).csv"
df = pd.read_csv(link_csv)
df
```

```
df. info ()
```

```
df.drop(columns=["Unnamed: 0",
                 "Unnamed: 0.2" ,
                 "Unnamed: 0.1"], inplace=True)
df
```

```
df. info()
```

```
df["ubicacion"].unique().size
```

```
df.drop(columns = ["ubicacion"], inplace = True)
df
```

```
import matplotlib.pyplot as plt
import seaborn as sns

# Histograma
plt.figure(
    figsize=(10,10),
    dpi = 100
)
sns.histplot(df['area'], kde=True)
plt.xlabel("area")
plt.ylabel("frecuencia")
plt.gca().spines[["top", "right"]].set_visible(False)
plt.show()
```
