```python

import pandas as pd
link_csv = "https://raw.githubusercontent.com/joshirdzdel-svg/estadisticainferencial2026A/refs/heads/main/Unidad_03/Practica_01/nasdaq_screener_1792638510409.csv"
df = pd.read_csv(link_csv)
df
```

```python
df.drop(columns=["Symbol",
                 "Name"], inplace=True)
df
```

```python

from numpy import column_stack
df = df. rename (columns = {
   "Last Sale":"Última venta",
   "Net Change":"Cambio neto",
   "% Change":"Cambio %",
   "Market Cap":"Capitalización de mercado",
   "Country":"País",
   "IPO Year":"Año de IPO",
   "Volume":"",
   "Sector":"",
   "Industry":""
})
df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Histograma
plt.figure(
    figsize=(5,3),
    dpi = 100
)
sns.histplot(df['carga_servidor'], kde=True)
plt.xlabel("Carga del servidor (%)")
plt.ylabel("Frecuencia")
plt.gca().spines[["top", "right"]].set_visible(False)
plt.show()
```
